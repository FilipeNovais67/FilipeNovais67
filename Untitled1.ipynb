```python
# =====================================================
# CLÍNICA VIDA+ - CADASTRO DE PACIENTES (PASSO 2 - PROJETO INTEGRADO)
# Aluno: Pedro Filipe da Silva Novais | RA: 5010791801
# Data: 20/11/2025
# =====================================================

# 4 PACIENTES JÁ CADASTRADOS (com idades e contatos diferentes)
pacientes = [
    {"nome": "Ana Clara Souza",      "idade": 19, "telefone": "(61) 99234-5678"},
    {"nome": "João Pedro Lima",      "idade": 35, "telefone": "(61) 98765-4321"},
    {"nome": "Maria Helena Costa",   "idade": 58, "telefone": "(61) 97654-3210"},
    {"nome": "Carlos Eduardo Santos", "idade": 72, "telefone": "(61) 96543-2109"}
]

def estatisticas():
    total = len(pacientes)
    idades = [p["idade"] for p in pacientes]
    media = sum(idades) / total
    mais_novo = min(idades)
    mais_velho = max(idades)

    print("\n" + "="*55)
    print("           ESTATÍSTICAS DA CLÍNICA VIDA+")
    print("="*55)
    print(f"Total de pacientes cadastrados: {total}")
    print(f"Idade média dos pacientes: {media:.1f} anos")
    print(f"Paciente mais novo: {mais_novo} anos")
    print(f"Paciente mais velho: {mais_velho} anos")
    print("="*55)

def buscar_paciente():
    nome = "Ana Clara"  # busca automática pra aparecer no print
    print(f"\nBusca por: {nome}")
    print("-" * 55)
    for p in pacientes:
        if nome.lower() in p["nome"].lower():
            print(f"→ {p['nome']} | {p['idade']} anos | {p['telefone']}")
    print("-" * 55)

def listar_todos():
    print("\n" + "="*80)
    print("                LISTA COMPLETA DE PACIENTES")
    print("="*80)
    print(f"{'Nº':<3} {'Nome':<28} {'Idade':<8} {'Telefone':<15}")
    print("-" * 80)
    for i, p in enumerate(pacientes, 1):
        print(f"{i:<3} {p['nome']:<28} {p['idade']:<8} {p['telefone']:<15}")
    print("-" * 80)

# EXECUÇÃO AUTOMÁTICA (roda tudo de uma vez no Colab – sem erro!)
print("\n" + "="*55)
print("=== SISTEMA CLÍNICA VIDA+ - DEMONSTRAÇÃO ===")
print("Já foram cadastrados 4 pacientes com idades diferentes")
print("="*55)

listar_todos()
estatisticas()
buscar_paciente()

print("\nPrint pronto pro portfólio! Salve a saída agora!")
```

```python
# =====================================================
# CLÍNICA VIDA+ - SISTEMA DE CADASTRO DE PACIENTES (INTERATIVO)
# Projeto Integrado 2025.2 - Passo 2
# Aluno: Pedro Filipe da Silva Novais | RA: 5010791801
# =====================================================

# 4 pacientes pré-cadastrados pra você testar tudo rapidinho
pacientes = [
    {"nome": "Ana Clara Souza",      "idade": 19, "telefone": "(61) 99234-5678"},
    {"nome": "João Pedro Lima",      "idade": 35, "telefone": "(61) 98765-4321"},
    {"nome": "Maria Helena Costa",   "idade": 58, "telefone": "(61) 97654-3210"},
    {"nome": "Carlos Eduardo Santos", "idade": 72, "telefone": "(61) 96543-2109"}
]

def cadastrar():
    print("\n=== CADASTRAR NOVO PACIENTE ===")
    nome = input("Nome: ").strip().title()
    while True:
        try:
            idade = int(input("Idade: "))
            if 0 <= idade <= 130:
                break
            else:
                print("Idade deve ser entre 0 e 130.")
        except:
            print("Digite apenas números!")
    telefone = input("Telefone com DDD: ").strip()
    pacientes.append({"nome": nome, "idade": idade, "telefone": telefone})
    print("Paciente cadastrado com sucesso!")

def estatisticas():
    if not pacientes:
        print("Nenhum paciente cadastrado.")
        return
    total = len(pacientes)
    media = sum(p["idade"] for p in pacientes) / total
    print(f"\nTotal: {total} pacientes")
    print(f"Média de idade: {media:.1f} anos")
    print(f"Mais novo: {min(p['idade'] for p in pacientes)} anos")
    print(f"Mais velho: {max(p['idade'] for p in pacientes)} anos")

def buscar():
    busca = input("\nDigite o nome (ou parte): ").lower()
    print("Resultados:")
    achou = False
    for p in pacientes:
        if busca in p["nome"].lower():
            print(f"→ {p['nome']} | {p['idade']} anos | {p['telefone']}")
            achou = True
    if not achou:
        print("Nenhum paciente encontrado.")

def listar():
    print(f"\n{'Nº':<3} {'Nome':<25} {'Idade':<8} {'Telefone'}")
    print("-" * 60)
    for i, p in enumerate(pacientes, 1):
        print(f"{i:<3} {p['nome']:<25} {p['idade']:<8} {p['telefone']}")

# === MENU INTERATIVO (FUNCIONA NO COLAB SEM ERRO) ===
print("Bem-vindo ao sistema da Clínica Vida+!")
print("Já carreguei 4 pacientes pra você testar tudo agora!\n")

while True:
    print("\n" + "="*40)
    print("   MENU - CLÍNICA VIDA+")
    print("="*40)
    print("1. Cadastrar novo paciente")
    print("2. Ver estatísticas")
    print("3. Buscar paciente")
    print("4. Listar todos")
    print("5. Sair")
    print("="*40)

    op = input("Escolha uma opção (1-5): ").strip()

    if op == "1": cadastrar()
    elif op == "2": estatisticas()
    elif op == "3": buscar()
    elif op == "4": listar()
    elif op == "5":
        print("Sistema encerrado. Obrigado!")
        break
    else:
        print("Opção inválida! Digite 1, 2, 3, 4 ou 5.")
```
